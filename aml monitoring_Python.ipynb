```python
#Data Importing
import pandas as pd
import numpy as np
INPUT = "aml_transaction_data.xlsx"
clients_df = pd.read_excel(INPUT, sheet_name="client data",usecols="A:L",engine="openpyxl")
transaction_df= pd.read_excel(INPUT, sheet_name="transaction data")
```

```python
#Data quality
assert clients_df["client_id"].is_unique
assert transaction_df["transaction_id"].is_unique
```

```python
#Data Cleaning-clients
clients_df = clients_df.dropna(axis=1, how="all")
clients_df['client_id'] = pd.to_numeric(clients_df['client_id']).astype('Int64')
# Also drop columns with the junk placeholder header pattern ("Column1", "Column2", ...)
import re
junk_cols = [c for c in clients_df.columns if re.match(r"^Column\d+$", str(c))]
clients_df = clients_df.drop(columns=junk_cols)

print(f"Remaining columns: {len(clients_df.columns)}")
print(clients_df.columns.tolist())
```

```python
#Data cleaning-Transaction
transaction_df["transaction_type"] = transaction_df["transaction_type"].str.upper()
transaction_df["timestamp"] = pd.to_datetime(transaction_df["timestamp"], unit="D", origin="1899-12-30")
transaction_df["transaction_date"] = transaction_df["timestamp"].dt.date
transaction_df["transaction_month"] = transaction_df["timestamp"].dt.to_period("M").astype(str)
transaction_df["transaction_hour"] = transaction_df["timestamp"].dt.hour
transaction_df["timestamp_display"] = transaction_df["timestamp"].dt.strftime("%Y-%m-%d %H:%M:%S")
transaction_df['amount'] = pd.to_numeric(transaction_df['amount'])
transaction_df['client_id'] = pd.to_numeric(transaction_df['client_id']).astype('Int64')
client_flag_cols = ['pep_flag','sanctions_flag','fatf_country_flag','ofac_country_flag','sectoral_sanctions_flag','ownership_opacity_score']
for col in client_flag_cols:
    clients_df[col] = pd.to_numeric(clients_df[col]).fillna(0).astype(int)
txn_flag_cols = ['ofac_match_flag','fatf_country_flag','structuring_pattern_flag','rapid_movement_flag','trade_mispricing_flag']
for col in txn_flag_cols:
    transaction_df[col] = pd.to_numeric(transaction_df[col]).fillna(0).astype(int)
```

```python
#Data Merging
analysis_df = transaction_df.merge(clients_df, on="client_id", how="left", validate="many_to_one", suffixes=("_txn","_client"))
```

```python
#Client level risk calculation
client_risk_flag_cols = ['pep_flag', 'sanctions_flag', 'fatf_country_flag_client', 'ofac_country_flag', 'sectoral_sanctions_flag','ownership_opacity_score']
analysis_df['client_flag_count'] = analysis_df[client_risk_flag_cols].sum(axis=1)
analysis_df['high_risk_client']= ((analysis_df["sector_risk"] == "High") |
    (analysis_df["client_flag_count"] > 0)).astype(int)
analysis_df['high_risk_client_transaction_share'] = (
    analysis_df["high_risk_client"].sum()
    / analysis_df["transaction_id"].count()
)
```

```python
#transaction level risk calulation
txn_risk_flag_cols = ['ofac_match_flag', 'fatf_country_flag_txn', 'structuring_pattern_flag', 'rapid_movement_flag', 'trade_mispricing_flag']
analysis_df['txn_flag_count'] = analysis_df[txn_risk_flag_cols].sum(axis=1)
analysis_df['suspicious_txn'] = (analysis_df['txn_flag_count'] > 0).astype(int)
analysis_df["txn_per_client"] = analysis_df.groupby("client_id")["client_id"].transform("count")
analysis_df['suspicious_transaction_count']= analysis_df["suspicious_txn"].sum()
analysis_df['suspicious_transaction_rate'] = (
    analysis_df["suspicious_txn"].sum() / analysis_df["transaction_id"].count()
) * 100
```

```python
# ---------------------------------------------------
# STEP 1: Aggregate transaction data to ONE ROW PER CLIENT
# ---------------------------------------------------

client_risk_df = (
    analysis_df
    .groupby("client_id", as_index=False)
    .agg(
        sector_risk=("sector_risk", "first"),
        client_flag_count=("client_flag_count", "max"),
        transaction_count=("transaction_id", "count"),
        transaction_flag_count=("txn_flag_count", "sum"),
        suspicious_transaction_count=("suspicious_txn", "sum"),
        total_value=("amount", "sum")
    )
)


# ---------------------------------------------------
# STEP 2: Sector Risk Score
# ---------------------------------------------------

client_risk_df["sector_risk_score"] = (
    client_risk_df["sector_risk"]
    .map({
        "High": 3,
        "Medium": 2,
        "Low": 1
    })
    .fillna(1)
)


# ---------------------------------------------------
# STEP 3: Client Flag Risk Score
# ---------------------------------------------------

client_risk_df["client_flag_risk_score"] = np.select(
    [
        client_risk_df["client_flag_count"] >= 5,
        client_risk_df["client_flag_count"] >= 2,
        client_risk_df["client_flag_count"] >= 1
    ],
    [
        3,
        2,
        1
    ],
    default=0
)


# ---------------------------------------------------
# STEP 4: Transaction Flag Risk Score
# ---------------------------------------------------

client_risk_df["txn_flag_risk_score"] = np.select(
    [
        client_risk_df["transaction_flag_count"] >= 5,
        client_risk_df["transaction_flag_count"] >= 2,
        client_risk_df["transaction_flag_count"] >= 1
    ],
    [
        3,
        2,
        1
    ],
    default=0
)


# ---------------------------------------------------
# STEP 5: Transaction Volume Risk Score
# ---------------------------------------------------

client_risk_df["txn_per_client_risk_score"] = np.select(
    [
        client_risk_df["transaction_count"] >= 40,
        client_risk_df["transaction_count"] >= 25
    ],
    [
        2,
        1
    ],
    default=0
)


# ---------------------------------------------------
# STEP 6: Composite Risk Score
# ---------------------------------------------------

client_risk_df["composite_risk_score"] = (
    client_risk_df["sector_risk_score"]
    + client_risk_df["client_flag_risk_score"]
    + client_risk_df["txn_flag_risk_score"]
    + client_risk_df["txn_per_client_risk_score"]
)

# ---------------------------------------------------
# STEP 7: Overall Risk Rating
# ---------------------------------------------------

client_risk_df["overall_risk_rating"] = np.select(
    [
        client_risk_df["composite_risk_score"] >= 5,
        client_risk_df["composite_risk_score"] >= 3
    ],
    [
        "High",
        "Medium"
    ],
    default="Low"
)


# ---------------------------------------------------
# STEP 8: Suspicious Transaction Rate
# ---------------------------------------------------

client_risk_df["suspicious_rate_pct"] = (
    client_risk_df["suspicious_transaction_count"]
    / client_risk_df["transaction_count"]
    * 100
).round(2)
```

```python
risk_summary = (
    client_risk_df["overall_risk_rating"]
    .value_counts()
    .rename_axis("overall_risk_rating")
    .reset_index(name="client_count")
)

print(risk_summary)
```
